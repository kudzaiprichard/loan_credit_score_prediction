# 02 · Cleaning & Feature Engineering
Turn the messy raw table into model-ready features. **Order matters:**
1. standardise formats (dedup cols, booleans→0/1, normalise the target text)
2. record **missingness** (before any imputation)
3. engineer rich features (amortization, affordability, logs, cyclical dates, scores)
4. learned **group/peer aggregates** (fit on train only)
5. impute → scale → encode (inside the Pipeline)

All logic lives in `src.cleaning` / `src.feature_engineering` / `src.preprocessing` so it runs identically at train **and** inference time.

```python
from src.data_loader import load_raw
from src.cleaning import clean_features, prepare_target
from src.feature_engineering import add_features
from src import config
raw = prepare_target(load_raw())
cleaned = clean_features(raw)
feats = add_features(cleaned)
print('after clean:', list(cleaned.columns))
print('row-wise engineered feature count:', feats.shape[1])
```

## Format standardisation
`is_employed` True/False → 1/0; the **target** is stripped, lower-cased and canonicalised (`'Did not default'`→0, `'Defaulted'`→1) with the positive class pinned explicitly (safer than a blind `LabelEncoder`, which orders labels alphabetically and could flip which class means 'default').

## Missingness as signal (does *not disclosing* predict default?)
We flag each field's missingness and count `total_missing_fields` **before** imputation, then test the hypothesis.

```python
import numpy as np, pandas as pd
y = raw[config.TARGET].to_numpy()
tab = pd.Series(y).groupby(feats['total_missing_fields'].values).mean().round(3)
print('default rate by #missing fields:'); print(tab)
print('correlation:', round(float(np.corrcoef(feats['total_missing_fields'].fillna(0), y)[0,1]), 4))
```

> Default rate rises with the number of undisclosed fields → **missingness is predictive**, so the flags + count earn their place in the model.

## Engineered feature families
Amortization (`estimated_monthly_payment`, `interest_cost_est`), affordability (`dti`, `loan_to_income`, `balance_to_income`), credit history, hand-coded risk scores, interaction terms, log transforms, and cyclical calendar features.

```python
fams = ['estimated_monthly_payment','interest_cost_ratio','dti','loan_to_income',
        'balance_to_loan','defaults_per_decade','job_risk_score','amount_x_rate',
        'log_loan_amount','month_sin']
feats[fams].describe().round(3)
```

## Learned group / peer-residual aggregates
Per `job` / `location` / `marital_status` we learn mean·median·std of income/amount/rate and a **deviation-from-peer** (`__dev`). Fit on the training fold only (leakage-safe), so they also work on a single applicant at inference — an improvement over the concat(train,test) Kaggle pattern.

```python
from src.preprocessing import GroupStatsEncoder
gse = GroupStatsEncoder().fit(feats, y)
added = [c for c in gse.transform(feats.head()).columns if '__' in c]
print('group features added:', len(added)); added
```

## Encoding strategy (no fake ordering)
- low-cardinality nominals (`gender, job, marital_status, age_bucket, rate_segment`) → **One-Hot**
- high-cardinality `location` (157 cities) → cross-fitted **TargetEncoder**
- numerics → median impute + StandardScaler (dynamic selector picks up new + group features; drift-prone `loan_year`/`loan_dayofyear` excluded).

```python
from src.preprocessing import make_pipeline, select_numeric_columns
from sklearn.linear_model import LogisticRegression
full = gse.transform(feats)
print('numeric features to model:', len(select_numeric_columns(full)))
make_pipeline(LogisticRegression(max_iter=200))
```
